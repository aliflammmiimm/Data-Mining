```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_data = 100

data_iot = {
    "ID_Perangkat": [f"IOT-{str(i).zfill(3)}" for i in range(1, n_data + 1)],
    "Jenis_Perangkat": np.random.choice(["Smartwatch", "Smart CCTV", "Smart Lamp", "Smart Speaker", "Termostat"], n_data),
    "Status_Koneksi": np.random.choice(["Online", "Offline"], n_data, p=[0.8, 0.2]),
    "Kualitas_Sinyal": np.random.choice(["Buruk", "Cukup", "Baik", "Sangat Baik"], n_data, p=[0.1, 0.2, 0.4, 0.3]),
    "Jumlah_Aktivasi_Harian": np.random.randint(1, 50, n_data),
    "Suhu_Perangkat_C": np.round(np.random.uniform(25.0, 60.0, n_data), 1),
    "Konsumsi_Daya_Wh": np.round(np.random.uniform(10.0, 300.0, n_data), 1)
}

df_iot = pd.DataFrame(data_iot)
df_iot.to_csv("dataset_iot_100.csv", index = False)
print("Dataset berhasil dibuat dan disimpan sebagai 'dataset_iot_100.csv'")
```

```python
rata_rata_aktivasi = df_iot["Jumlah_Aktivasi_Harian"].mean()
rata_rata_suhu = df_iot["Suhu_Perangkat_C"].mean()
rata_rata_daya = df_iot["Konsumsi_Daya_Wh"].mean()

print("Rata-rata Aktivitas Harian: ", round(rata_rata_aktivasi, 2))
print("Rata-rata Suhu Perangkat: ", round(rata_rata_suhu, 2))
print("Rata-rata Konsumsi Daya: ", round(rata_rata_daya, 2))
```

```python
median_aktivasi = df_iot["Jumlah_Aktivasi_Harian"].median()
median_suhu = df_iot["Suhu_Perangkat_C"].median()
median_daya = df_iot["Konsumsi_Daya_Wh"].median()

print("Median Aktivitas Harian: ", round(median_aktivasi, 2))
print("Median Suhu Perangkat: ", round(median_suhu, 2))
print("Median Konsumsi Daya: ", round(median_daya, 2))
```

```python
statistik_ringkas = df_iot[
    [
        "Jumlah_Aktivasi_Harian",
        "Suhu_Perangkat_C",
        "Konsumsi_Daya_Wh"
    ]
].agg(
    [
        "mean",
        "median",
        "std",
        "min",
        "max"
    ]
)

statistik_ringkas
```

```python
fitur_kategorikal = [
    "Jenis_Perangkat",
    "Status_Koneksi",
    "Kualitas_Sinyal"
]

for kolom in fitur_kategorikal:
    print("=" * 60)
    print("ATRIBUT:", kolom)
    frekuensi = df_iot[kolom].value_counts()
    persentase = (
        df_iot[kolom]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
    )

    ringkasan = pd.DataFrame({
        "Frekuensi": frekuensi,
        "Persentase": persentase
    })

    print(ringkasan)
    print()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize = (8, 5))
sns.histplot(
    data = df_iot,
    x = "Suhu_Perangkat_C",
    bins = 10,
    kde = True
)

plt.title("Distribusi Suhu Perangkat")
plt.xlabel("Suhu Perangkat (Celcius)")
plt.ylabel("Frekuensi")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize = (8, 5))
sns.histplot(
    data = df_iot,
    x="Konsumsi_Daya_Wh",
    bins = 10,
    kde = True
)

plt.title("Distribusi Konsumsi Daya")
plt.xlabel("Konsumsi Daya (Wh)")
plt.ylabel("Frekuensi")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize = (8, 5))
sns.countplot(
    data = df_iot,
    x = "Jenis_Perangkat"
)

plt.title("Distribusi Jenis Perangkat IoT")
plt.xlabel("Jenis Perangkat")
plt.ylabel("Jumlah Perangkat")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize = (8, 5))
sns.countplot(
    data = df_iot,
    x = "Kualitas_Sinyal",
    order = ["Buruk", "Cukup", "Baik", "Sangat Baik"]
)

plt.title("Distribusi Kualitas Sinyal")
plt.xlabel("Kualitas Sinyal")
plt.ylabel("Jumlah Perangkat")

plt.show()
```

```python
import matplotlib.pyplot as plt

status_counts = df_iot["Status_Koneksi"].value_counts()

plt.figure(figsize = (7, 7))
plt.pie(
    status_counts.values,
    labels = status_counts.index,
    autopct = "%1.1f%%",
    startangle = 90
)

plt.title("Proporsi Status Koneksi")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize = (8, 6))
sns.scatterplot(
    data = df_iot,
    x = "Jumlah_Aktivasi_Harian",
    y = "Konsumsi_Daya_Wh"
)

plt.title("Hubungan Jumlah Aktivasi Harian dengan Konsumsi Daya")
plt.xlabel("Jumlah Aktivasi Harian")
plt.ylabel("Konsumsi Daya (Wh)")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize = (8, 6))
sns.regplot(
    data = df_iot,
    x = "Suhu_Perangkat_C",
    y = "Konsumsi_Daya_Wh"
)

plt.title("Hubungan Suhu Perangkat dengan Konsumsi Daya")
plt.xlabel("Suhu Perangkat (Celcius)")
plt.ylabel("Konsumsi Daya (Wh)")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fitur_korelasi = [
    "Jumlah_Aktivasi_Harian",
    "Suhu_Perangkat_C",
    "Konsumsi_Daya_Wh"
]

matriks_korelasi = df_iot[fitur_korelasi].corr()

plt.figure(figsize = (8, 6))
sns.heatmap(
    matriks_korelasi,
    annot = True,
    fmt = ".2f"
)

plt.title("Heatmap Korelasi Atribut Numerik")

plt.show()
```
