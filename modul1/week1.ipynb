```python
import pandas as pd
import folium
from folium.plugins import HeatMap
data_lokasi = {
    "Lokasi": [
        "Titik 01",
        "Titik 02",
        "Titik 03",
        "Titik 04",
        "Titik 05",
        "Titik 06",
        "Titik 07",
        "Titik 08",
        "Titik 09",
        "Titik 10",
        "Titik 11",
        "Titik 12",
        "Titik 13",
        "Titik 14",
        "Titik 15",
        "Titik 16",
        "Titik 17",
        "Titik 18",
        "Titik 19",
        "Titik 20"
    ],
    "Latitude": [
        -5.1477,
        -5.1510,
        -5.1545,
        -5.1580,
        -5.1615,
        -5.1650,
        -5.1685,
        -5.1720,
        -5.1755,
        -5.1790,
        -5.1825,
        -5.1860,
        -5.1895,
        -5.1930,
        -5.1965,
        -5.2000,
        -5.2035,
        -5.2070,
        -5.2105,
        -5.2140
    ],
    "Longitude": [
        119.4327,
        119.4370,
        119.4410,
        119.4450,
        119.4490,
        119.4530,
        119.4570,
        119.4610,
        119.4650,
        119.4690,
        119.4730,
        119.4770,
        119.4810,
        119.4850,
        119.4890,
        119.4930,
        119.4970,
        119.5010,
        119.5050,
        119.5090
    ],
    "Aktivitas": [
        15,
        25,
        35,
        55,
        70,
        90,
        85,
        65,
        45,
        30,
        20,
        40,
        60,
        80,
        95,
        75,
        50,
        35,
        25,
        15
    ]
}

df_lokasi = pd.DataFrame(data_lokasi)

# Membuat peta
peta = folium.Map(
    location = [-5.18, 119.47],
    zoom_start = 12
)

# Menyiapkan data heatmap
heat_data = df_lokasi[
    [
        "Latitude",
        "Longitude",
        "Aktivitas"
    ]
].values.tolist()

# Menambahkan heatmap
HeatMap(
    heat_data,
    radius=25,
    blur=20,
    min_opacity=0.3
).add_to(peta)

# Menambahkan titik observasi
for _, baris in df_lokasi.iterrows():
    folium.CircleMarker(
        location=[
            baris["Latitude"],
            baris["Longitude"]
        ],
        radius=5,
        popup=(
            f"Lokasi: {baris['Lokasi']}<br>"
            f"Aktivitas: {baris['Aktivitas']}"
        ),
        tooltip=baris["Lokasi"],
        fill=True
    ).add_to(peta)
    
peta.save('peta.html')
```
